```python


conn = sqlite3.connect(
    r"C:\Users\midhlaj muhmmd\OneDrive\Desktop\airtel-telecom-analytics\data\raw\airtel.db"
)

query = """
SELECT
    COUNT(DISTINCT region) AS distinct_regions,
    SUM(region IS NULL) AS nulls,
    SUM(TRIM(region) = '') AS blanks,
    ROUND(
        100.0 * AVG(
            region IS NULL OR TRIM(region) = ''
        ), 2
    ) AS pct_missing,
    MIN(region) AS smallest,
    MAX(region) AS largest
FROM circles;
"""

df = pd.read_sql_query(query, conn)
display(df)

conn.close()
```

```python


# Connect to the Airtel database
conn = sqlite3.connect(
    r"C:\Users\midhlaj muhmmd\OneDrive\Desktop\airtel-telecom-analytics\data\raw\airtel.db"
)

# SQL query
query = """
SELECT
    txn_type,
    COUNT(*) AS payments,
    COUNT(DISTINCT customer_id) AS subs,
    ROUND(AVG(amount_inr), 2) AS avg_inr
FROM recharges
GROUP BY txn_type;
"""

# Execute query and display results
df = pd.read_sql_query(query, conn)
display(df)

conn.close()
```

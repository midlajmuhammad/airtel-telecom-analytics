```python
def run_sql_file(path):
    with open(path, 'r', encoding='utf-8') as f:
        script = f.read()
    con.executescript(script)
    con.commit()
    print('ran', path)

run_sql_file(r'C:\Users\midhlaj muhmmd\OneDrive\Desktop\airtel-telecom-analytics\sql\practice\00_test.sql')

q('SELECT COUNT(*) AS n FROM customers;')
```
